# Simple Bot
- Define state structure with a list of Human Message objects.
- Initialize a GPT-4o model using LangChain's ChatOpenAI
- Sending and handling different types
- Building and compailing the graph of the agents.

## Main Goal:
How to integrate LLMs in our graphs

```python
from typing import TypedDict, List
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, START, END
from dotenv import load_dotenv
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
class AgentState(TypedDict):
    messages: List[HumanMessage]

llm = ChatOpenAI(model="gpt-4o", temperature=0.0)
```

```python
def process(state: AgentState)->AgentState:
    """
    Process the messages in the state and return the updated state.
    """
    response = llm.invoke(state['messages'])
    print(f"\nAI: {response.content}")
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node('process', process)
graph.add_edge(START, 'process')
graph.add_edge('process', END)
agent = graph.compile()
```

```python
user_input = input("Enter : ")
agent.invoke({'messages':[HumanMessage(content = user_input)]})
```

## ChatBot Type

```python
user_input = input("Enter : ")
while user_input.lower() != 'exit':
    agent.invoke({'messages':[HumanMessage(content=user_input)]})
    user_input = input("Enter : ")
```
